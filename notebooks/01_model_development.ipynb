# Resume Screening Agent — Model Development Notebook

This notebook walks through the resume parsing, feature extraction, similarity scoring, and model training pipeline.

```python
import sys
sys.path.insert(0, '../src')

from parse_resume import parse_resumes_from_folder
from similarity import SimilarityScorer
from rank_candidates import ResumeRanker

jd_path = '../data/sample_job_descriptions/software_engineer_jd.txt'
with open(jd_path, 'r', encoding='utf-8') as f:
    jd = f.read()

resumes = parse_resumes_from_folder('../data/sample_resumes')
print(f"Parsed {len(resumes)} resumes")
```

```python
scorer = SimilarityScorer()
sims = scorer.compute_batch_semantic_similarity(jd, list(resumes.values()))
for name, sim in zip(resumes.keys(), sims):
    print(f"{name}: {sim:.3f}")
```

```python
ranker = ResumeRanker(models_dir='../models')
results = ranker.rank(jd, resumes)
results[['rank', 'filename', 'score', 'skill_match_ratio', 'experience_fit']]
```
